```python
#Problema 1: Programa que imprime la constante de Mandelung para el cloruro de sodio.
import numpy as np
import math as mt

L=100
M=0

for i in range (-L,L+1):
 for j in range (-L,L+1):
  for k in range (-L,L+1):

   if i==j==k==0:
    continue
 d=np.sqrt(i**2+j**2+k**2)

if (i+j+k)%2==0:
   signo=1
else:
    signo=-1

M=M+signo/d 

print("La constante de Mandelung es:", M)
```
